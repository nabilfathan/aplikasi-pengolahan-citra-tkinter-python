# Membuat Aplikasi GUI Desktop Menggunakan Python dengan Tkinter
**Tkinter** merupakan built-in library Python untuk membuat ***Graphical User Interfaces (GUIs)***. Tkinter membungkus (*wrapper*) modul *[Tcl/Tk GUI toolkit](https://www.tcl-lang.org/)*, sehingga developer Python dapat dengan mudah dan sederhana membangun desktop applications. Library ini mendukung layout management, event handling and customization, membuatnya ideal untuk rapid GUI development di Python.

Mengapa kita perlu Tkinter?
- Menyediakan 'built-in' library dan mudah digunakan untuk membuat Aplikasi GUI (Graphical User Interface) di Python.
- Menawarkan beragam 'widgets' seperti `buttons`, `labels`, `text boxes`, `menus` dan lain-lain.
- Menghilang kebutuhan external GUI frameworks karena tkinter di bundle bersama Python.
- Support event-driven programming, membuatnya cocok untuk user interfaces yang responsive

## A. Membuat Aplikasi Tkinter Pertama
Untuk membuat aplikasi Tkinter di Python, ikuti langkah dasar berikut:<br>
1. Impor modul tkinter: Impor modul `tkinter` yang diperlukan untuk membuat komponen GUI.
2. Buat window utama atau biasanya diberi nama dengan `root` : Inisialisasi `root` aplikasi menggunakan kelas `Tk()`.
3. Atur properti window : Kita dapat mengatur properti seperti judul (`title`) dan ukuran window (`geometry`).
4. Tambahkan `widget` ke `root`: Tambahkan widget seperti `button`, `label`, `entry`, dan lain-lain untuk merancang antarmuka.
5. Susun `widget`: Gunakan pengelola tata letak seperti `pack()`, `grid()` atau `place()` untuk menata widget di dalam jendela.
6. Pasang pemicu (`handler`) pada widget: Lampirkan `event handler` pada widget agar merespons interaksi pengguna (mis. `klik tombol`).

Ada dua metode utama yang perlu diingat saat membuat aplikasi GUI dengan Tkinter:

1. `Tk()`
Untuk membuat `main window` atau `root` di Tkinter, gunakan kelas `Tk()`.

    Contoh sintaks: `root = Tk(screenName=None, baseName=None, className='Tk', useTk=1)`

    Parameter:
    - `screenName` (opsional): Menentukan tampilan (umumnya untuk sistem Unix dengan beberapa layar).
    - `baseName` (opsional): Menentukan nama dasar aplikasi (default adalah nama skrip).
    - `className` (opsional): Nama kelas jendela (berguna untuk styling dan pengaturan window manager).
    - `useTk` (opsional): Boolean yang menentukan apakah subsistem Tk diinisialisasi (biasanya default `1`).

2. `mainloop()` : Metode `mainloop()` untuk menjalankan aplikasi sampai selesai (close). 
    
    Ini adalah loop tak hingga yang menjaga aplikasi tetap berjalan, menunggu dan memproses event sampai jendela ditutup.

## C. Membuat Menu pada window utama atau window aplikasi
Cobalah buat **menu** pada **window utama atau root** dengan menambahkan `widget Menu`, `menu_name = Menu(master, option=value)`

1. Untuk pembuatan aplikasi yang baik dan modular, sebuah aplikasi sebaiknya di kemas dalam sebuah Class
ketik kode berikut: lalu jalankan (run)

```python
import tkinter as tk
from tkinter import *

# definisikan class aplikasi
# -----------------------------------------awal Class------------------------------------------
class ImageProcessingApp:
    
    #definisikan App class constructor
    def __init__(self, root):

        # inisialisasi window aplikasi
        self.root = root
        self.root.title("Aplikasi Pengolahan Citra Digital")
        self.root.geometry("800x600")

        # tambahkan menu (menuBar) ke window aplikasi(root)  
        self.menuBar = Menu(self.root)
        self.root.config(menu=self.menuBar)

        # buat item menu     
        self.menuFile = Menu(self.menuBar)
        self.menuEdit = Menu(self.menuBar)
        self.menuOpen = Menu(self.menuFile)
        self.menuSave = Menu(self.menuFile)
        self.menuExit = Menu(self.menuFile)
        
        # Menambahkan item kedalam menu:
        # - add_cascade   --> menu dengan sub_menu,
        # - add_command   --> menu tanpa sub_menu (tombol),
        # - add_separator --> pemisah
        
        # tambah item ke menuBar
        self.menuBar.add_cascade(label='File', menu=self.menuFile)
        self.menuBar.add_cascade(label='Edit', menu=self.menuEdit)

        # tambah item ke menuFile
        self.menuFile.add_command(label='Open', command=self.menuOpen_click)
        self.menuFile.add_command(label='Save', command=self.menuSave_click)
        self.menuFile.add_separator()
        self.menuFile.add_command(label='Exit', command=self.menuExit_click)

    #------akhir constructor-----------------

    #------awal fungsi untuk setiap item menu-----

    def menuOpen_click(self):
        print("Menu Open diklik")
    
    def menuSave_click(self):
        print("Menu Save diklik")
    
    def menuExit_click(self):
        print("Menu Exit diklik")
        self.root.destroy()  # menutup window aplikasi

    #------akhir fungsi untuk setiap item menu-----

# -----------------------------------akhir Class------------------------------------

#----Awal Program Utama (main)-----------
if __name__ == "__main__":
    # buat object window
    root = tk.Tk()
    
    # buat instance dari class aplikasi, dan masukkan object window nya sebagai parameter
    app = ImageProcessingApp(root)

    # jalankan window aplikasi
    root.mainloop()

#----Akhir Program Utama (main)----------- 
```

## D. Membuka, Menampilkan, dan Menyimpan Gambar
Sekarang menu **File** kita buat benar-benar berfungsi dengan library **Pillow** (`PIL`):
1. `Open` : pilih file gambar memakai `filedialog.askopenfilename()`, lalu buka dengan `Image.open()`.
2. Tampilkan gambar pada `Canvas`. Tkinter tidak bisa menampilkan objek PIL secara langsung, jadi gambar diubah dulu dengan `ImageTk.PhotoImage()`.
3. `Save` : simpan gambar hasil olahan memakai `filedialog.asksaveasfilename()` dan `image.save()`.
4. `Exit` : tutup aplikasi.

> Simpan referensi `PhotoImage` pada atribut class (`self.photo`), kalau tidak gambar akan hilang karena dibersihkan *garbage collector*.

```python
import tkinter as tk
from tkinter import filedialog, messagebox
from PIL import Image, ImageTk


class ImageProcessingApp:

    def __init__(self, root):
        self.root = root
        self.root.title("Aplikasi Pengolahan Citra Digital")
        self.root.geometry("800x600")

        self.image = None
        self.photo = None

        self.status = tk.Label(self.root, text="Silakan buka gambar: File > Open", anchor="w")
        self.status.pack(side=tk.BOTTOM, fill=tk.X)

        self.canvas = tk.Canvas(self.root, bg="gray20")
        self.canvas.pack(fill=tk.BOTH, expand=True)
        self.canvas.bind("<Configure>", self._on_resize)

        self.menuBar = tk.Menu(self.root)
        self.root.config(menu=self.menuBar)

        self.menuFile = tk.Menu(self.menuBar, tearoff=0)
        self.menuBar.add_cascade(label="File", menu=self.menuFile)
        self.menuFile.add_command(label="Open", command=self.menuOpen_click)
        self.menuFile.add_command(label="Save", command=self.menuSave_click)
        self.menuFile.add_separator()
        self.menuFile.add_command(label="Exit", command=self.menuExit_click)

    def _on_resize(self, event):
        self.tampilkan()

    def tampilkan(self):
        self.canvas.delete("all")
        if self.image is None:
            return

        cw = max(self.canvas.winfo_width(), 1)
        ch = max(self.canvas.winfo_height(), 1)

        tampil = self.image.copy()
        tampil.thumbnail((cw, ch))

        self.photo = ImageTk.PhotoImage(tampil)
        self.canvas.create_image(cw // 2, ch // 2, image=self.photo)
        self.status.config(text=f"Ukuran: {self.image.width} x {self.image.height} px")

    def menuOpen_click(self):
        jenis_file = [
            ("Gambar", "*.png *.jpg *.jpeg *.bmp *.gif *.tiff"),
            ("Semua file", "*.*"),
        ]
        path = filedialog.askopenfilename(filetypes=jenis_file)
        if path:
            with Image.open(path) as img:
                self.image = img.convert("RGB")
            self.tampilkan()

    def menuSave_click(self):
        if self.image is None:
            messagebox.showwarning("Peringatan", "Belum ada gambar.")
            return

        format_simpan = [
            ("PNG", "*.png"),
            ("JPEG", "*.jpg"),
            ("BMP", "*.bmp"),
        ]
        path = filedialog.asksaveasfilename(defaultextension=".png", filetypes=format_simpan)
        if path:
            self.image.save(path)

    def menuExit_click(self):
        self.root.destroy()


if __name__ == "__main__":
    root = tk.Tk()
    app = ImageProcessingApp(root)
    root.mainloop()
```

## E. Operasi Pengolahan Citra sebagai Fungsi, Dipanggil dari Menu
Sesuai soal, setiap operasi dibuat sebagai **fungsi Python** yang menerima gambar PIL dan mengembalikan gambar baru (`op_...`). Fungsi-fungsi ini dipanggil dari menu melalui method `terapkan()`.

| Menu | Operasi | Fungsi Pillow |
|---|---|---|
| **Edit** | Undo, Kembalikan Gambar Asli | riwayat gambar |
| **Warna** | Grayscale, Negatif, Sepia | `ImageOps.grayscale`, `ImageOps.invert`, `ImageOps.colorize` |
| **Intensitas** | Brightness, Contrast, Gamma, Threshold (Biner) | `ImageEnhance`, `point()` |
| **Geometri** | Flip Horizontal/Vertikal, Rotate 90/180/270/Bebas, Resize | `ImageOps.mirror/flip`, `rotate`, `resize` |
| **Filter** | Blur, Gaussian Blur, Median, Sharpen, Edge Detection, Contour, Emboss | `ImageFilter` |
| **Histogram** | Tampilkan Histogram, Equalization, Auto Contrast | `histogram()`, `ImageOps.equalize/autocontrast` |

Jalankan sel di bawah ini, lalu uji dengan File > Open dan coba setiap menu.

```python
import tkinter as tk
from tkinter import filedialog, messagebox, simpledialog
from PIL import Image, ImageTk, ImageOps, ImageFilter, ImageEnhance

# =====================================================================
# FUNGSI-FUNGSI OPERASI (input: gambar PIL RGB, output: gambar PIL RGB)
# =====================================================================

# ---- Warna
def op_grayscale(img):
    return ImageOps.grayscale(img).convert("RGB")

def op_negatif(img):
    return ImageOps.invert(img)

def op_sepia(img):
    gray = ImageOps.grayscale(img)
    return ImageOps.colorize(gray, black="#2b1a0a", white="#ffeccc", mid="#a07850")

# ---- Intensitas
def op_brightness(img, faktor):
    return ImageEnhance.Brightness(img).enhance(faktor)

def op_contrast(img, faktor):
    return ImageEnhance.Contrast(img).enhance(faktor)

def op_gamma(img, gamma):
    tabel = [int(255 * ((i / 255) ** (1 / gamma))) for i in range(256)]
    return img.point(tabel * 3)          # tabel yang sama untuk R, G, B

def op_threshold(img, batas):
    biner = ImageOps.grayscale(img).point(lambda p: 255 if p > batas else 0)
    return biner.convert("RGB")

# ---- Geometri
def op_flip_horizontal(img):
    return ImageOps.mirror(img)

def op_flip_vertikal(img):
    return ImageOps.flip(img)

def op_rotate(img, sudut):
    return img.rotate(sudut, expand=True)   # berlawanan arah jarum jam

def op_resize(img, persen):
    w, h = max(int(img.width * persen / 100), 1), max(int(img.height * persen / 100), 1)
    return img.resize((w, h), Image.LANCZOS)

# ---- Filter
def op_blur(img):
    return img.filter(ImageFilter.BLUR)

def op_gaussian_blur(img, radius):
    return img.filter(ImageFilter.GaussianBlur(radius))

def op_median(img):
    return img.filter(ImageFilter.MedianFilter(3))

def op_sharpen(img):
    return img.filter(ImageFilter.SHARPEN)

def op_edge(img):
    return img.filter(ImageFilter.FIND_EDGES)

def op_contour(img):
    return img.filter(ImageFilter.CONTOUR)

def op_emboss(img):
    return img.filter(ImageFilter.EMBOSS)

# ---- Histogram
def op_equalization(img):
    return ImageOps.equalize(img)

def op_autocontrast(img):
    return ImageOps.autocontrast(img)


# =====================================================================
# CLASS APLIKASI
# =====================================================================
class ImageProcessingApp:

    def __init__(self, root):
        self.root = root
        self.root.title("Aplikasi Pengolahan Citra Digital")
        self.root.geometry("800x600")

        self.original = None   # gambar asli
        self.image = None      # gambar hasil olahan
        self.history = []      # riwayat untuk Undo
        self.photo = None      # referensi PhotoImage

        self.status = tk.Label(self.root, text="Silakan buka gambar: File > Open", anchor="w")
        self.status.pack(side=tk.BOTTOM, fill=tk.X)
        self.canvas = tk.Canvas(self.root, bg="gray20")
        self.canvas.pack(fill=tk.BOTH, expand=True)
        self.canvas.bind("<Configure>", lambda e: self.tampilkan())

        self.buat_menu()

    # ------------------------------ MENU ------------------------------
    def buat_menu(self):
        self.menuBar = tk.Menu(self.root)
        self.root.config(menu=self.menuBar)

        def submenu(nama):
            m = tk.Menu(self.menuBar, tearoff=0)
            self.menuBar.add_cascade(label=nama, menu=m)
            return m

        # File
        m = submenu("File")
        m.add_command(label="Open", command=self.menuOpen_click)
        m.add_command(label="Save", command=self.menuSave_click)
        m.add_separator()
        m.add_command(label="Exit", command=self.menuExit_click)

        # Edit
        m = submenu("Edit")
        m.add_command(label="Undo", command=self.menuUndo_click)
        m.add_command(label="Kembalikan Gambar Asli", command=self.menuReset_click)

        # Warna
        m = submenu("Warna")
        m.add_command(label="Grayscale", command=lambda: self.terapkan(op_grayscale))
        m.add_command(label="Negatif", command=lambda: self.terapkan(op_negatif))
        m.add_command(label="Sepia", command=lambda: self.terapkan(op_sepia))

        # Intensitas
        m = submenu("Intensitas")
        m.add_command(label="Brightness...", command=lambda: self.terapkan_dialog(
            op_brightness, "Brightness", "Nilai 0-300 (100 = asli):", 100, 0, 300, lambda v: v / 100))
        m.add_command(label="Contrast...", command=lambda: self.terapkan_dialog(
            op_contrast, "Contrast", "Nilai 0-300 (100 = asli):", 100, 0, 300, lambda v: v / 100))
        m.add_command(label="Gamma...", command=lambda: self.terapkan_dialog(
            op_gamma, "Gamma", "Nilai 10-300 (100 = 1.0):", 100, 10, 300, lambda v: v / 100))
        m.add_command(label="Threshold (Biner)...", command=lambda: self.terapkan_dialog(
            op_threshold, "Threshold", "Batas 0-255:", 128, 0, 255))

        # Geometri
        m = submenu("Geometri")
        m.add_command(label="Flip Horizontal", command=lambda: self.terapkan(op_flip_horizontal))
        m.add_command(label="Flip Vertikal", command=lambda: self.terapkan(op_flip_vertikal))
        m.add_separator()
        m.add_command(label="Rotate 90", command=lambda: self.terapkan(op_rotate, 90))
        m.add_command(label="Rotate 180", command=lambda: self.terapkan(op_rotate, 180))
        m.add_command(label="Rotate 270", command=lambda: self.terapkan(op_rotate, 270))
        m.add_command(label="Rotate Bebas...", command=lambda: self.terapkan_dialog(
            op_rotate, "Rotate", "Sudut (-360 s/d 360):", 45, -360, 360))
        m.add_separator()
        m.add_command(label="Resize...", command=lambda: self.terapkan_dialog(
            op_resize, "Resize", "Skala (%) 1-500:", 50, 1, 500))

        # Filter
        m = submenu("Filter")
        m.add_command(label="Blur", command=lambda: self.terapkan(op_blur))
        m.add_command(label="Gaussian Blur...", command=lambda: self.terapkan_dialog(
            op_gaussian_blur, "Gaussian Blur", "Radius 1-20:", 3, 1, 20))
        m.add_command(label="Median", command=lambda: self.terapkan(op_median))
        m.add_command(label="Sharpen", command=lambda: self.terapkan(op_sharpen))
        m.add_separator()
        m.add_command(label="Edge Detection", command=lambda: self.terapkan(op_edge))
        m.add_command(label="Contour", command=lambda: self.terapkan(op_contour))
        m.add_command(label="Emboss", command=lambda: self.terapkan(op_emboss))

        # Histogram
        m = submenu("Histogram")
        m.add_command(label="Tampilkan Histogram", command=self.menuHistogram_click)
        m.add_command(label="Equalization", command=lambda: self.terapkan(op_equalization))
        m.add_command(label="Auto Contrast", command=lambda: self.terapkan(op_autocontrast))

    # ------------------------- FUNGSI BANTU ---------------------------
    def cek_gambar(self):
        if self.image is None:
            messagebox.showwarning("Peringatan", "Buka gambar dulu: File > Open")
            return False
        return True

    def tampilkan(self):
        self.canvas.delete("all")
        if self.image is None:
            return
        cw, ch = max(self.canvas.winfo_width(), 1), max(self.canvas.winfo_height(), 1)
        tampil = self.image.copy()
        tampil.thumbnail((cw, ch))
        self.photo = ImageTk.PhotoImage(tampil)
        self.canvas.create_image(cw // 2, ch // 2, image=self.photo)
        self.status.config(text=f"Ukuran: {self.image.width} x {self.image.height} px")

    def terapkan(self, fungsi, *args):
        """Panggil fungsi operasi pada gambar saat ini."""
        if not self.cek_gambar():
            return
        self.history.append(self.image)
        self.history = self.history[-20:]
        self.image = fungsi(self.image, *args)
        self.tampilkan()

    def terapkan_dialog(self, fungsi, judul, prompt, awal, minimum, maksimum, konversi=lambda v: v):
        """Minta nilai parameter ke pengguna, lalu panggil fungsi operasi."""
        if not self.cek_gambar():
            return
        nilai = simpledialog.askinteger(judul, prompt, initialvalue=awal,
                                        minvalue=minimum, maxvalue=maksimum, parent=self.root)
        if nilai is not None:
            self.terapkan(fungsi, konversi(nilai))

    # ----------------------------- FILE -------------------------------
    def menuOpen_click(self):
        path = filedialog.askopenfilename(
            filetypes=[("Gambar", "*.png *.jpg *.jpeg *.bmp *.gif *.tiff"), ("Semua file", "*.*")])
        if path:
            self.original = Image.open(path).convert("RGB")
            self.image = self.original
            self.history = []
            self.tampilkan()

    def menuSave_click(self):
        if not self.cek_gambar():
            return
        path = filedialog.asksaveasfilename(
            defaultextension=".png",
            filetypes=[("PNG", "*.png"), ("JPEG", "*.jpg"), ("BMP", "*.bmp")])
        if path:
            self.image.save(path)

    def menuExit_click(self):
        self.root.destroy()

    # ----------------------------- EDIT -------------------------------
    def menuUndo_click(self):
        if self.history:
            self.image = self.history.pop()
            self.tampilkan()

    def menuReset_click(self):
        if self.original is not None:
            self.history.append(self.image)
            self.image = self.original
            self.tampilkan()

    # --------------------------- HISTOGRAM ----------------------------
    def menuHistogram_click(self):
        if not self.cek_gambar():
            return
        h = self.image.histogram()      # 768 nilai: 256 R + 256 G + 256 B
        win = tk.Toplevel(self.root)
        win.title("Histogram")
        c = tk.Canvas(win, width=276, height=230, bg="white")
        c.pack()
        maks = max(h) or 1
        for k, warna in enumerate(("red", "green", "blue")):
            titik = []
            for i in range(256):
                titik += [10 + i, 210 - h[k * 256 + i] / maks * 200]
            c.create_line(*titik, fill=warna)
        c.create_line(10, 210, 266, 210)   # sumbu x
        c.create_text(10, 220, text="0", anchor="w")
        c.create_text(266, 220, text="255", anchor="e")


# =====================================================================
# PROGRAM UTAMA
# =====================================================================
if __name__ == "__main__":
    root = tk.Tk()
    app = ImageProcessingApp(root)
    root.mainloop()
```
